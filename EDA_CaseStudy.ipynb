```python
import matplotlib.pyplot as plt
```

```python
plt.hist(data['Age'],bins=15,color='red')
plt.xlabel("Age")
plt.ylabel("count")
plt.title("Age Distribution")
plt.show()
```

```python
#diaplay the count of bars in histograms
counts, edges, bars = plt.hist(data['Age']) 
plt.title("Age Distribution")
plt.bar_label(bars)
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
#to print count variables
print(counts)
print(bars)
```

```python
#bar plot representation
data['Sex'].value_counts().plot(kind='bar')
```

```python
data['Pclass'].value_counts().plot(kind='barh',color = 'yellow')
```

```python
plt.bar(data['Survived'], data['Sex'])
plt.xlabel("Survived rate")
plt.ylabel("Gender")
plt.title("Gender Survived Distribution")
plt.show()
```

```python
data['Survived'].value_counts()
```

```python
index = data['Survived'].value_counts().keys().tolist()
print(index)
labels = data['Survived'].value_counts().values.tolist()
print(labels)
```

```python
plt.pie(x= labels, labels= ['Dead', 'Alive'] ,colors = ['red','yellow'], autopct=("%.1f%%"))
plt.show()
```

```python
plt.pie(x= [577,314], labels= ['M', 'F'], colors=['purple','pink'], autopct=("%.1f%%"), shadow=True)
plt.legend()
plt.show()
```

```python
#pairwise relationship ---->scatterplot()
plt.scatter(x= data['Age'],
            y = data['Fare'], color= "purple")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare Distribution")
plt.show()
```
